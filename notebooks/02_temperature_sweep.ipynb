# 02 - Coolant Temperature Sweep
Radiator area and mass collapse as ~T^-4 when coolant temperature rises from 40 C to 1000 C.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
plt.rcParams.update({'figure.dpi': 110, 'font.size': 11})
from models.high_temperature_model.run import run
df = run(); df
```

```python
fig, ax = plt.subplots(1,2, figsize=(11,4))
ax[0].plot(df['coolant_C'], df['area_m2']/1e3, 'o-')
ax[0].set_xlabel('Coolant T [C]'); ax[0].set_ylabel('Area [1000 m^2]')
ax[0].set_title('Radiator area vs temperature')
ax[1].plot(df['coolant_C'], df['mass_reduction_x'], 's-', color='crimson')
ax[1].set_xlabel('Coolant T [C]'); ax[1].set_ylabel('Mass reduction (x)')
ax[1].set_title('Mass reduction vs 40 C baseline')
plt.tight_layout(); plt.show()
```
