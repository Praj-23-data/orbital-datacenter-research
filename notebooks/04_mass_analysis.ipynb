# 04 - Mass Analysis
Radiator mass across radiator technologies at a common 250 C operating point.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
plt.rcParams.update({'figure.dpi': 110, 'font.size': 11})
from models.droplet_radiator_model.run import run as drun
from models.liquid_sheet_model.run import run as srun
d = drun(); s = srun()
masses = {'Panel': d['panel_mass_kg'], 'Liquid sheet': s['sensitivity']['total_mass_kg'].min(), 'Droplet': d['droplet_mass_kg']}
pd.Series(masses, name='mass_kg')
```

```python
ser = pd.Series({'Panel': d['panel_mass_kg'], 'Liquid sheet': s['sensitivity']['total_mass_kg'].min(), 'Droplet': d['droplet_mass_kg']})
ax = (ser/1e3).plot.bar(color=['steelblue','seagreen','indianred'])
ax.set_ylabel('Mass [t]'); ax.set_title('Radiator mass by concept @250 C')
plt.tight_layout(); plt.show()
```
